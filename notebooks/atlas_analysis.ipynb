```python
import numpy as np
import matplotlib.pyplot as plt

atlases = ['A2009', 'DKT', 'APARC', 'WM', 'ASEG']
models = ['SVM', 'SVM TR', 'FNN(1)', 'FNN(3)']

# Test set (verified)
mae_test = np.array([
    [5.92, 5.33, 5.69, 5.81],   # A2009
    [6.41, 5.72, 6.26, 0],   # DKT
    [6.43, 6.01, 6.15, 0],   # APARC
    [6.61, 5.68, 6.27, 0],   # WM
    [5.78, 5.24, 5.66, 0]    # ASEG
])

std_test = np.array([
    [0.22, 0.27, 0.29, 0.36],
    [0.14, 0.21, 0.34, 0],
    [0.20, 0.15, 0.16, 0],
    [0.27, 0.23, 0.20, 0],
    [0.12, 0.40, 0.15, 0]
])

# Holdout set (verified)
mae_holdout = np.array([
    [6.15, 5.48, 6.07, 0],
    [6.42, 6.08, 6.32, 0],
    [6.67, 6.39, 6.47, 0],
    [7.26, 7.08, 6.81, 0],
    [6.02, 5.51, 5.74, 0]
])

std_holdout = np.array([
    [0.03, 0.07, 0.06, 0],
    [0.08, 0.38, 0.10, 0],
    [0.06, 0.34, 0.04, 0],
    [0.10, 0.41, 0.20, 0],
    [0.04, 0.07, 0.13, 0]
])

colors = ['#F4A3C2', '#C38DDB', '#6FCF97', "#7979E0"]
```

```python
x = np.arange(len(atlases))
x = np.arange(len(atlases)) * 1.5
width = 0.25

fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# ---- Test set ----
for i, model in enumerate(models):
    axes[0].bar(
        x + i*width,
        mae_test[:, i],
        width,
        yerr=std_test[:, i],
        capsize=4,
        label=model,
        color=colors[i]
    )

axes[0].set_title('Zbiór testowy', fontsize=14)
axes[0].set_ylabel('MAE', fontsize=14)
axes[0].set_xticks(x + width)
axes[0].set_xticklabels(atlases, fontsize=14)
axes[0].grid(axis='y', linestyle='--', alpha=0.6)

# ---- Holdout set ----
for i, model in enumerate(models):
    axes[1].bar(
        x + i*width,
        mae_holdout[:, i],
        width,
        yerr=std_holdout[:, i],
        capsize=4,
        label=model,
        color=colors[i]
    )

axes[1].set_title('Zbiór holdout', fontsize=14)
axes[1].set_xticks(x + width)
axes[1].set_xticklabels(atlases, fontsize=14)
axes[1].grid(axis='y', linestyle='--', alpha=0.6)

# Wspólna legenda
fig.legend(models, loc='upper center', ncol=3, frameon=False, fontsize=14)

plt.tight_layout(rect=[0, 0, 1, 0.92])
plt.show()
```
