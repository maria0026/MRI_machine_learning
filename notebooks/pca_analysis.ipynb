```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
pca_components = np.array([30, 35, 40, 50, 70, 80, 100, 110, 120, 130, 140, 150, 180, 250])
mae_mean = np.array([5.07, 5.01, 5.03, 5.16, 4.92, 5.1, 4.89, 4.96, 4.93, 4.81, 5.05, 5.17, 5.11, 5.09])
mae_std = np.array([0.19, 0.23, 0.27, 0.28, 0.15, 0.14, 0.14, 0.13, 0.22, 0.19, 0.26, 0.27, 0.17, 0.22])

# Wykres
plt.errorbar(
    pca_components,
    mae_mean,
    yerr=mae_std,
    fmt='o-',
    linewidth=2,
    capsize=5
)

plt.xlabel('Liczba komponentów PCA', fontsize=12)
plt.ylabel('MAE', fontsize=12)
#make bigger numbers on x and y axis
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
#make more ticks on x axis
plt.xticks(np.arange(25, 260, 20))
#plt.title('MAE w zależności od liczby komponentów PCA dla modelu SVM')
plt.grid(True)
plt.show()
```
